## Data Structures

### Tuples

A `tuple` is used to represent an **immutable** sequence of objects. Here you find an example of a tuple:

````python
t = (0, 1, "hello", "world")
print(t)
````

You can access the 3rd value of a tuple like this:

````python
t[2]
````

What happens if you try to change a value of a tuple?

````python
t[2] = "No can do!"
````

Tuples are usually used to return multiple values in a function:

````python
def sum_and_diff(a, b):
    return (a + b, a - b)

sum_and_diff(1, -1)
````

### Lists

A `list` is used to represent a **mutable** sequence of objects. Here you find an example of a list:

````python
l = [0, 1, "hello", "world"]
print(l)
````

You can access or change a value of the list by index like this:

````python
l[2] = "happy"
print(l)
````

You can append a new value to a `list` like this:

````python
l.append("!")
print(l)
````

You can insert a value at any position you want like this:

````python
l.insert(2, "you should delete until here!")
print(l)
````

You can delete a value by index with `pop` like this:

````python
l.pop(1)
print(l)
l.pop()
print(l)
````

You can remove a value by value like this:

````python
l.remove("world")
print(l)
````

### Sets

A `set` is an unordered collection of unique, hashable objects (such as numbers and strings). Here you find an example of a set:

````python
s = {"hello", "world", 1, 0, 100.85471}
print(s)
````

You can add a new value to a set like this:

````python
s.add(2.58745)
print(s)
````

Sets cannot contain duplicates. For example, what happens if you add the character `!` 100 times to the set `s`?

````python
for _ in range(100):
    s.add("!")

print(s)
````

Note that, unlike lists, in sets the order is not preserved.

You can remove an item from a list using `pop()`. If no index is specified, it removes the last item. For a set, `pop()` removes an arbitrary item.

````python
s.pop()
````

Why does `set.pop()` seem to follow the order shown by `print()`?
You may notice that `set.pop()` often removes elements in the same order in which they appear when the set is printed:

````python
s = {1, 0, "hello", "world", 2.58745, 100.85471, "!"}

print(s)  # The displayed order may vary.

while s:
    print(s.pop())
````

In CPython, both `print(s)` and `s.pop()` scan the set's internal storage. For this freshly created set, they may encounter the elements in the same order. This matching order is not guaranteed by Python.

However, this does not mean that sets have an order. The order is simply a result of how Python currently stores the elements internally. It may change if you use different elements, modify the set, restart Python (because of hash randomisation for strings), or use a different Python implementation or version.

Therefore, `set.pop()` should be described as removing an arbitrary element, not as removing elements in the order shown by `print()`.

What does `pop()` return?

`pop()` removes an item from the set and returns the item that it removed:

````python
s = {"apple", "banana", "cherry"}

item = s.pop()

print(item)  # The item removed from the set
print(s)     # The set after removal
````

You can remove an item by value (not index) like this:

````python
s = {1, 0, 2, "hello", "world"}

s.remove(2)
print(s)
````

Calling `s.remove(2)` again without recreating the set raises a `KeyError`, because `2` has already been removed:

````python
s.remove(2)
````

When you are not sure whether the item you want to remove exists in the set, use `discard()` to remove it:

````python
s.discard(2)
print(s)
````

This does not raise an error if the item does not exist. If the item does exist, it removes it:

````python
s.discard(1)
print(s)
````

### Dictionaries

A `dictionary` is used to represent a series of value objects associated to a set of key objects. Here you find an example of a dictionary:

````python
d = {0:0, 1:0, "hello":1, "world":100}
print(d)
````

You can access or update the value associated to a key like this:

````python
d[1] = "olleh"
print(d)
````

You can add a key like this:

````python
d["!"] = "great"
print(d)
````

You can remove a key by using pop:

````python
d.pop("!")
````

You can access the keys like this:

````python
d.keys()
````

And, the values like this:

````python
d.values()
````

To access both keys and values as `(key, value)` tuples, use `items()`. This returns a dictionary view, which you can iterate over:

````python
d.items()
````

# Exercise 19: From a notebook to a Python script

Inside your existing `CEGE0096` folder, create `Exercise_19/main.py`. Adapt your solution to Exercise 16:

1. Write a function `count_groceries(items)` that takes a list of grocery items and returns a dictionary mapping each item to its quantity.
2. In `main()`, create the grocery list from Exercise 14 (five apples, two kiwis and one bag of cereals), call your function, then print each quantity and item on a separate line.
3. Call `main()` using `if __name__ == '__main__':`, as in Week 2.

Keep counting inside `count_groceries()` and printing inside `main()`. The counting function should use its `items` parameter, so it also works with a different list or an empty list.

````python
def count_groceries(items):
    counts = {}
    for item in items:
        if item not in counts:
            counts[item] = 0
        counts[item] = counts[item] + 1
    return counts


def main():
    items = ["apple", "apple", "apple", "apple", "apple", "kiwi", "kiwi", "cereals"]
    counts = count_groceries(items)
    for item, quantity in counts.items():
        print(quantity, item)


if __name__ == '__main__':
    main()
````
